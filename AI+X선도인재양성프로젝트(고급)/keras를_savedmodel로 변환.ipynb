```python
import tensorflow as tf
import os
from tensorflow.keras.models import load_model

# MobileNetV3 전용 전처리가 로드 시 필요할 수 있으므로 import (필수)
from tensorflow.keras.applications.mobilenet_v3 import preprocess_input

############################
plant_name = "호접란"
############################

KERAS_MODEL_PATH = f"/content/drive/MyDrive/식물을 부탁해/kjw/mobilenetv3_log/{plant_name}_mobilenetv3_large_best(final).keras"
# SavedModel 경로를 지정할 때는 확장자가 없어야 합니다.
SAVEDMODEL_PATH = KERAS_MODEL_PATH.replace(".keras", "_savedmodel_FINAL_K3_V2")

# --- 커스텀 객체는 그대로 유지 ---
class FixedInputLayer(tf.keras.layers.InputLayer):
    """Keras 버전 업데이트로 인한 InputLayer 호환성 문제를 해결하는 클래스"""
    def __init__(self, **kwargs):
        if 'batch_shape' in kwargs:
            kwargs['batch_input_shape'] = kwargs.pop('batch_shape')
        super().__init__(**kwargs)

CUSTOM_OBJECTS = {
    'InputLayer': FixedInputLayer,
    'DTypePolicy': tf.keras.mixed_precision.DTypePolicy
}
# -----------------------------

print(f"1. .keras 모델 로드 시작: {KERAS_MODEL_PATH}")

try:
    # custom_objects를 사용하여 .keras 모델을 로드
    keras_model = load_model(
        KERAS_MODEL_PATH,
        compile=False,
        custom_objects=CUSTOM_OBJECTS
    )
    print("✅ .keras 모델 로드 성공.")

    # --- 입력 텐서 이름 확인 ---
    input_name = 'input_layer'
    try:
        input_tensor = keras_model.inputs[0]
        raw_name = input_tensor.name.split(':')[0]
        input_name = raw_name
        print(f"✅ 모델의 입력 텐서 이름 확인: **'{input_name}'** (원래 이름: {input_tensor.name})")
    except Exception as e:
        print(f"⚠️ 모델 입력 이름 추출 실패 ({e}). 기본값 '{input_name}'을 사용합니다.")


    # ⭐️ 핵심: serving_default 시그니처를 위한 tf.function 정의 ⭐️
    # input_signature의 이름을 추출된 이름으로 사용합니다.
    @tf.function(input_signature=[
        tf.TensorSpec(
            shape=[None, 224, 224, 3], # 배치 크기(None), 이미지 크기(224, 224), 채널(3)
            dtype=tf.float32,
            name=input_name
        )
    ])
    def serving_fn(input_tensor):
        # Keras 모델의 추론을 호출합니다.
        # 출력 딕셔너리 포맷을 맞춥니다.
        return {'output_0': keras_model(input_tensor)}


    print(f"\n2. SavedModel 포맷으로 저장 시작 (Keras Model.export() API 사용): {SAVEDMODEL_PATH}")

    # ⭐️ 최종 문제 해결 부분: Keras Model 객체의 .export() 메서드를 사용 ⭐️
    # Keras 3에서 모델을 SavedModel로 변환하면서 시그니처를 지정하는 가장 안정적인 방법입니다.
    # Keras 모델 객체에 signatures 딕셔너리를 전달하여 내보냅니다.

    keras_model.export(
        SAVEDMODEL_PATH,
        signatures={'serving_default': serving_fn} # signatures를 명시적으로 전달
    )

    print("\n--- 🌟 최종 변환 완료 (Keras Model.export) 🌟 ---")
    print(f"✅ 새로운 SavedModel은 다음 경로에 '폴더' 형태로 저장되었습니다:")
    print(f"   -> {SAVEDMODEL_PATH}")
    print("\n💡 이 모델은 'serving_default' 시그니처를 포함하며 서버에서 안정적으로 로드될 것입니다.")

except Exception as e:
    print("\n--- ❌ 모델 로드/변환 최종 실패 ---")
    print(f"최종 오류: {e}")
    print("⚠️ 이 방법이 Keras 3에서 가장 표준적인 방식이므로, 여기서 오류가 발생한다면 환경(TF/Keras 버전)을 확인해야 합니다.")
```
