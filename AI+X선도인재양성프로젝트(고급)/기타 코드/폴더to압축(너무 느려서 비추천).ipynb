## 드라이브 내의 폴더를 압축폴더로 만들고 싶다면 아래의 코드를 실행(파일명과 경로 수정 필수!!!!!)

```python
import os
import shutil

# 압축할 폴더 경로 (원본 폴더)
file_name = "보스턴고사리_dataset" # 파일 이름(폴더명은 여기만 바꾸면 됨)
src_folder = f"/content/drive/MyDrive/AI+X고급/식물을 부탁해/kjw/{file_name}"  # 압축할 폴더 경로 /content/drive/MyDrive/AI+X고급/식물을 부탁해/kjw는 지우고 자신의 경로를 넣으세용.. 그대로 쓰면 제 폴더로 갑니다...

# 압축 결과 파일 경로 (/content 밑에 생성됨)
zip_path = f"/content/{file_name}"

# 압축 실행 (부모 폴더를 root_dir, 실제 폴더명을 base_dir로 지정)
shutil.make_archive(
    zip_path,
    'zip',
    root_dir=os.path.dirname(src_folder),
    base_dir=os.path.basename(src_folder)
)

print("✅ 압축 완료:", zip_path + ".zip")
```

```python
# 생성된 zip을 구글 드라이브에 복사
dest_path = "/content/drive/MyDrive/AI+X고급/식물을 부탁해/kjw/" # 압축폴더를 저장할 위치
!cp "{zip_path}.zip" "{dest_path}"

print("✅ 드라이브 복사 완료:", dest_path)
```
